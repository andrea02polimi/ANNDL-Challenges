```python
import os
# --- FIX MEMORIA CUDA ---
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

import cv2
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
import gc
from sklearn.preprocessing import LabelEncoder
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torchvision import transforms, models
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, confusion_matrix
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from torchvision.transforms import AutoAugment, AutoAugmentPolicy 
from torch.cuda.amp import autocast, GradScaler

# --- 1. CONFIGURAZIONE "CAMPIONE" ---
# Usiamo la stessa configurazione che ha dato 0.3808, ma applichiamola alla B1
SEEDS = [42, 2024, 777]  
N_FOLDS = 5
IMG_SIZE = 256        # Ottimo anche per B1
BATCH_SIZE = 32       # La B1 è leggera, dovrebbe reggere
EPOCHS = 20           # Punto di equilibrio perfetto
LR = 1e-4 
MIXUP_ALPHA = 0.4
MODEL_NAME = "b1"     # Identificativo per i file

def find_paths():
    base_train = '/kaggle/input/dataset/archive'
    if not os.path.exists(base_train): base_train = '/kaggle/input/dataset'
    test_img_path = None
    for root, dirs, files in os.walk('/kaggle/input'):
        if root.endswith("test_data/images"):
            test_img_path = root; break
    if test_img_path is None:
        test_img_path = '/kaggle/input/dataset/archive (1)/dataset challenge 2 dl /test_data/images'
    return {
        'train_img': os.path.join(base_train, 'images'),
        'train_mask': os.path.join(base_train, 'masks'),
        'labels': os.path.join(base_train, 'train_labels.csv'),
        'test_img': test_img_path, 
        'test_mask': test_img_path.replace('images', 'masks')
    }

PATHS = find_paths()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
PRECOMPUTED_DIR_TRAIN = '/kaggle/working/train_precomputed'
PRECOMPUTED_DIR_TEST = '/kaggle/working/test_precomputed'

# --- 2. UTILS ---
def seed_everything(seed):
    random.seed(seed)
    os.environ['PYTHONHASHSEED'] = str(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
    print(f"🌱 Seed set to: {seed}")

class ReinhardNormalizer:
    def __init__(self):
        self.target_mean = np.array([148.60, 169.30, 105.47])
        self.target_std = np.array([41.56, 11.23, 14.39])
    def __call__(self, img_rgb):
        try:
            img_lab = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2LAB).astype("float32")
            img_mean = np.mean(img_lab, axis=(0,1))
            img_std = np.std(img_lab, axis=(0,1)) + 1e-5
            img_lab = (img_lab - img_mean) / img_std * self.target_std + self.target_mean
            img_lab = np.clip(img_lab, 0, 255).astype("uint8")
            return cv2.cvtColor(img_lab, cv2.COLOR_LAB2RGB)
        except: return img_rgb

def get_bbox_from_mask(mask):
    if np.sum(mask) == 0: return None
    rows, cols = np.where(mask > 0)
    return np.min(rows), np.max(rows), np.min(cols), np.max(cols)

def preprocess_and_save_batch(df, src_img_dir, src_mask_dir, output_dir, is_test=False):
    if not os.path.exists(output_dir): os.makedirs(output_dir)
    reinhard = ReinhardNormalizer()
    valid_files = []
    
    for _, row in tqdm(df.iterrows(), total=len(df), desc="Precomputing"):
        img_name = row['filename'] if is_test else row['sample_index']
        mask_name = img_name.replace("img", "mask") if "img" in img_name else img_name.replace(".png", "_mask.png")
        img_p = os.path.join(src_img_dir, img_name)
        mask_p = os.path.join(src_mask_dir, mask_name)
        out_p = os.path.join(output_dir, img_name)
        
        if os.path.exists(out_p): valid_files.append(img_name); continue
        try:
            image = cv2.imread(img_p)
            if image is None: continue
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
            if os.path.exists(mask_p):
                mask = cv2.imread(mask_p, cv2.IMREAD_GRAYSCALE)
                if mask is not None:
                    bbox = get_bbox_from_mask(mask)
                    if bbox: image = image[bbox[0]:bbox[1]+1, bbox[2]:bbox[3]+1]
            image = reinhard(image)
            Image.fromarray(image).resize((IMG_SIZE, IMG_SIZE), Image.Resampling.LANCZOS).save(out_p, "PNG")
            valid_files.append(img_name)
        except: continue
    return valid_files

class TissueDataset(Dataset):
    def __init__(self, dataframe, precomputed_dir, transform=None, is_test=False):
        self.df = dataframe
        self.precomputed_dir = precomputed_dir 
        self.transform = transform
        self.is_test = is_test
    def __len__(self): return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_name = row['filename'] if self.is_test else row['sample_index']
        img_path = os.path.join(self.precomputed_dir, img_name)
        try: image = Image.open(img_path).convert("RGB")
        except: image = Image.new('RGB', (IMG_SIZE, IMG_SIZE))
        if self.transform: image = self.transform(image)
        if self.is_test: return image, img_name
        else: return image, torch.tensor(row['label_encoded'], dtype=torch.long)

def get_transforms():
    train = transforms.Compose([
        transforms.RandomResizedCrop(IMG_SIZE, scale=(0.8, 1.0)), 
        AutoAugment(policy=AutoAugmentPolicy.IMAGENET),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
    val = transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])
    return train, val

def mixup_data(x, y, alpha=1.0):
    if alpha > 0: lam = np.random.beta(alpha, alpha)
    else: lam = 1
    index = torch.randperm(x.size(0)).to(DEVICE)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)

def get_model(num_classes):
    # --- UPGRADE: EfficientNet-B1 ---
    # Ha più capacità della B0 ma non è enorme come la B3
    model = models.efficientnet_b1(weights=models.EfficientNet_B1_Weights.DEFAULT)
    model.classifier[1] = nn.Linear(model.classifier[1].in_features, num_classes)
    return model.to(DEVICE)

# --- 3. TRAINING ENGINE (CORRETTO) ---
def train_one_seed(seed, df_final, le, precomputed_dir):
    seed_everything(seed)
    skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
    
    # CORREZIONE: Inizializza la lista QUI, prima del ciclo dei fold
    histories = [] 
    
    for fold, (t_idx, v_idx) in enumerate(skf.split(df_final, df_final['label_encoded'])):
        print(f"\n--- Seed {seed} | Fold {fold+1}/{N_FOLDS} (EffNet-B1) ---")
        train_df = df_final.iloc[t_idx].reset_index(drop=True)
        val_df = df_final.iloc[v_idx].reset_index(drop=True)
        
        counts = train_df['label_encoded'].value_counts().sort_index().values
        weights = 1. / counts
        s_weights = [weights[l] for l in train_df['label_encoded']]
        sampler = WeightedRandomSampler(s_weights, len(s_weights))
        
        train_loader = DataLoader(TissueDataset(train_df, precomputed_dir, get_transforms()[0]), 
                                  batch_size=BATCH_SIZE, sampler=sampler, num_workers=2)
        val_loader = DataLoader(TissueDataset(val_df, precomputed_dir, get_transforms()[1]), 
                                batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
        
        model = get_model(len(le.classes_))
        optimizer = optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-2)
        criterion = nn.CrossEntropyLoss()
        scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS)
        scaler = GradScaler()
        
        best_f1 = 0.0
        fold_hist = {'t_loss': [], 'v_loss': [], 'v_f1': []}
        
        for epoch in range(EPOCHS):
            model.train()
            t_loss = 0
            
            for imgs, lbls in tqdm(train_loader, desc=f"Ep {epoch+1}", leave=False):
                imgs, lbls = imgs.to(DEVICE), lbls.to(DEVICE)
                optimizer.zero_grad()
                with autocast():
                    imgs, ta, tb, lam = mixup_data(imgs, lbls, MIXUP_ALPHA)
                    out = model(imgs)
                    loss = mixup_criterion(criterion, out, ta, tb, lam)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
                t_loss += loss.item()
            
            scheduler.step()
            
            # Validation
            model.eval()
            v_loss = 0
            v_preds, v_targets = [], []
            with torch.no_grad():
                for imgs, lbls in val_loader:
                    imgs, lbls = imgs.to(DEVICE), lbls.to(DEVICE)
                    with autocast():
                        out = model(imgs)
                        loss = criterion(out, lbls)
                    v_loss += loss.item()
                    v_preds.extend(out.argmax(1).cpu().numpy())
                    v_targets.extend(lbls.cpu().numpy())
            
            t_loss /= len(train_loader)
            v_loss /= len(val_loader)
            f1 = f1_score(v_targets, v_preds, average='macro')
            
            print(f"Ep {epoch+1}: T_Loss {t_loss:.4f} | V_Loss {v_loss:.4f} | Val F1 {f1:.4f}")
            fold_hist['t_loss'].append(t_loss)
            fold_hist['v_loss'].append(v_loss)
            fold_hist['v_f1'].append(f1)

            if f1 > best_f1:
                best_f1 = f1
                torch.save(model.state_dict(), f"{MODEL_NAME}_seed{seed}_fold{fold}.pth")
        
        print(f"🏆 Best F1 Fold {fold+1}: {best_f1:.4f}")
        
        # Aggiungi lo storico di questo fold alla lista generale
        histories.append(fold_hist)
        
        del model, optimizer, scaler
        torch.cuda.empty_cache()
        gc.collect()
        
    return histories

# --- 4. PLOTTING ---
def plot_seed_results(histories, seed_name):
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    for i, h in enumerate(histories):
        axes[0].plot(h['t_loss'], linestyle='--', alpha=0.4, label=f'F{i+1}')
        axes[1].plot(h['v_f1'], linewidth=2, label=f'F{i+1}')
    axes[0].set_title(f'Loss (Seed {seed_name})'); axes[1].set_title(f'F1 (Seed {seed_name})')
    plt.tight_layout(); plt.show()

# --- 5. SUBMISSION ---
def generate_submission(df, le):
    print("\n--- Generating Submission with Seed Averaging ---")
    test_files = sorted([f for f in os.listdir(PATHS['test_img']) if f.endswith('.png')])
    test_df = pd.DataFrame({'filename': test_files})
    
    valid_test = preprocess_and_save_batch(test_df, PATHS['test_img'], PATHS['test_mask'], PRECOMPUTED_DIR_TEST, is_test=True)
    test_df = test_df[test_df['filename'].isin(valid_test)].reset_index(drop=True)
    test_df['label_encoded'] = 0 
    
    test_ds = TissueDataset(test_df, PRECOMPUTED_DIR_TEST, get_transforms()[1], is_test=True)
    loader = DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
    
    all_models = []
    print("\nLoading models...")
    for seed in SEEDS:
        for fold in range(N_FOLDS):
            path = f"{MODEL_NAME}_seed{seed}_fold{fold}.pth"
            if os.path.exists(path):
                m = get_model(len(le.classes_))
                m.load_state_dict(torch.load(path))
                m.eval()
                all_models.append(m)
    
    print(f"📦 Loaded {len(all_models)} models (EfficientNet-B1).")
    
    preds, fnames = [], []
    with torch.no_grad():
        for imgs, names in tqdm(loader, desc="Inference"):
            imgs = imgs.to(DEVICE)
            inputs = [imgs, torch.flip(imgs, [3]), torch.flip(imgs, [2]), torch.rot90(imgs, 1, [2, 3])]
            avg_probs = torch.zeros((imgs.size(0), len(le.classes_))).to(DEVICE)
            
            for m in all_models:
                for x in inputs:
                    with autocast():
                        avg_probs += torch.softmax(m(x), dim=1)
            
            avg_probs /= (len(all_models) * 4)
            preds.extend(le.inverse_transform(avg_probs.argmax(1).cpu().numpy()))
            fnames.extend(names)
            
    pd.DataFrame({'sample_index': fnames, 'label': preds}).to_csv('submission_b1_ensemble.csv', index=False)
    print("✅ Done! Saved 'submission_b1_ensemble.csv'.")

if __name__ == "__main__":
    df = pd.read_csv(PATHS['labels'])
    valid_mask = df['sample_index'].apply(lambda x: os.path.exists(os.path.join(PATHS['train_img'], x)))
    df = df[valid_mask].reset_index(drop=True)
    le = LabelEncoder()
    df['label_encoded'] = le.fit_transform(df['label'])
    
    valid_train = preprocess_and_save_batch(df, PATHS['train_img'], PATHS['train_mask'], PRECOMPUTED_DIR_TRAIN)
    df_final = df[df['sample_index'].isin(valid_train)].reset_index(drop=True)
    
    for seed in SEEDS:
        h = train_one_seed(seed, df_final, le, PRECOMPUTED_DIR_TRAIN)
        plot_seed_results(h, seed)
        
    generate_submission(df_final, le)
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import f1_score, confusion_matrix
from sklearn.preprocessing import LabelEncoder
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
from torch.cuda.amp import autocast

# --- CONFIGURAZIONE ---
SEEDS = [42, 2024, 777]
N_FOLDS = 5
IMG_SIZE = 256
BATCH_SIZE = 64  # Più alto per inferenza veloce
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
PRECOMPUTED_DIR_TRAIN = '/kaggle/working/train_precomputed' 

# Funzione Paths (la stessa usata in training)
def find_paths():
    base_train = '/kaggle/input/dataset/archive'
    if not os.path.exists(base_train): base_train = '/kaggle/input/dataset'
    return {
        'train_img': os.path.join(base_train, 'images'),
        'train_labels': os.path.join(base_train, 'train_labels.csv')
    }

PATHS = find_paths()

# --- DATASET & TRANSFORMS ---
class TissueDataset(Dataset):
    def __init__(self, dataframe, precomputed_dir, transform=None):
        self.df = dataframe
        self.precomputed_dir = precomputed_dir 
        self.transform = transform

    def __len__(self): return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        # Usa 'sample_index' per caricare le immagini di train
        img_name = row['sample_index']
        img_path = os.path.join(self.precomputed_dir, img_name)
        
        try: image = Image.open(img_path).convert("RGB")
        except: image = Image.new('RGB', (IMG_SIZE, IMG_SIZE))
        
        if self.transform: image = self.transform(image)
        return image, torch.tensor(row['label_encoded'], dtype=torch.long)

def get_val_transforms():
    return transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])

# --- MODELLO B1 ---
def get_model_b1(num_classes):
    model = models.efficientnet_b1(weights=None)
    model.classifier[1] = nn.Linear(model.classifier[1].in_features, num_classes)
    return model.to(DEVICE)

# --- FUNZIONE DI PLOT ---
def plot_confusion_matrix(targets, preds, classes, title):
    cm = confusion_matrix(targets, preds)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=classes, yticklabels=classes)
    plt.title(title)
    plt.ylabel('True Label')
    plt.xlabel('Predicted Label')
    plt.show()

# --- ANALISI ---
def analyze_b1_models(le, df_final):
    print("🚀 Inizio analisi modelli EfficientNet-B1...")
    
    # Per ogni Seed, accumuliamo predizioni e target
    for seed in SEEDS:
        print(f"\n--- Analisi Seed {seed} ---")
        skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=seed)
        
        seed_targets = []
        seed_preds = []
        
        for fold, (_, v_idx) in enumerate(skf.split(df_final, df_final['label_encoded'])):
            val_df = df_final.iloc[v_idx].reset_index(drop=True)
            val_loader = DataLoader(TissueDataset(val_df, PRECOMPUTED_DIR_TRAIN, get_val_transforms()), 
                                    batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
            
            # Carica il modello B1 specifico per questo seed/fold
            model_path = f"b1_seed{seed}_fold{fold}.pth"
            if not os.path.exists(model_path):
                print(f"⚠️ Modello {model_path} non trovato! Salto...")
                continue
                
            model = get_model_b1(len(le.classes_))
            model.load_state_dict(torch.load(model_path, map_location=DEVICE))
            model.eval()
            
            # Inferenza
            with torch.no_grad():
                for imgs, lbls in val_loader:
                    imgs = imgs.to(DEVICE)
                    # Inferenza standard (senza TTA per velocità di analisi, o aggiungila se vuoi precisione assoluta)
                    outputs = model(imgs)
                    preds = outputs.argmax(1).cpu().numpy()
                    
                    seed_preds.extend(preds)
                    seed_targets.extend(lbls.numpy())

        # Calcola metriche per questo Seed
        if seed_preds:
            seed_f1 = f1_score(seed_targets, seed_preds, average='macro')
            print(f"📊 Seed {seed} - OOF Macro F1: {seed_f1:.4f}")
            
            # Plotta la matrice
            plot_confusion_matrix(seed_targets, seed_preds, le.classes_, f"Confusion Matrix - EffNet B1 (Seed {seed})")

# --- MAIN ---
if __name__ == "__main__":
    # Carica Dati e Label Encoder
    df = pd.read_csv(PATHS['train_labels'])
    
    # Filtra solo immagini esistenti (come nel training)
    # Nota: Assumiamo che la cartella train_precomputed esista già dal training precedente
    # Se non hai più il df_final in memoria, lo ricreiamo:
    valid_mask = df['sample_index'].apply(lambda x: os.path.exists(os.path.join(PRECOMPUTED_DIR_TRAIN, x)))
    df_final = df[valid_mask].reset_index(drop=True)
    
    le = LabelEncoder()
    df_final['label_encoded'] = le.fit_transform(df_final['label'])
    
    # Esegui Analisi
    analyze_b1_models(le, df_final)
```

```python
import os
import cv2
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms, models
from sklearn.preprocessing import LabelEncoder
from PIL import Image
from tqdm import tqdm
from torch.cuda.amp import autocast

# --- 1. CONFIGURAZIONE ENSEMBLE ---
SEEDS = [42, 2024, 777]
N_FOLDS = 5
IMG_SIZE = 256
BATCH_SIZE = 32
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# PESI DELL'ENSEMBLE (Somma = 1.0)
# Diamo più peso alla B0 perché ha uno score singolo più alto (0.3808)
WEIGHT_B0 = 0.60
WEIGHT_B1 = 0.40

# Cartelle
PRECOMPUTED_DIR_TEST = '/kaggle/working/test_precomputed_ensemble'

def find_paths():
    base_train = '/kaggle/input/dataset/archive'
    if not os.path.exists(base_train): base_train = '/kaggle/input/dataset'
    
    test_img_path = None
    for root, dirs, files in os.walk('/kaggle/input'):
        if root.endswith("test_data/images"):
            test_img_path = root; break
    if test_img_path is None:
        test_img_path = '/kaggle/input/dataset/archive (1)/dataset challenge 2 dl /test_data/images'
        
    return {
        'train_labels': os.path.join(base_train, 'train_labels.csv'),
        'test_img': test_img_path, 
        'test_mask': test_img_path.replace('images', 'masks')
    }

PATHS = find_paths()

# --- 2. UTILS & DATASET ---
class ReinhardNormalizer:
    def __init__(self):
        self.target_mean = np.array([148.60, 169.30, 105.47])
        self.target_std = np.array([41.56, 11.23, 14.39])
    def __call__(self, img_rgb):
        try:
            img_lab = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2LAB).astype("float32")
            img_mean = np.mean(img_lab, axis=(0,1))
            img_std = np.std(img_lab, axis=(0,1)) + 1e-5
            img_lab = (img_lab - img_mean) / img_std * self.target_std + self.target_mean
            img_lab = np.clip(img_lab, 0, 255).astype("uint8")
            return cv2.cvtColor(img_lab, cv2.COLOR_LAB2RGB)
        except: return img_rgb

def get_bbox_from_mask(mask):
    if np.sum(mask) == 0: return None
    rows, cols = np.where(mask > 0)
    return np.min(rows), np.max(rows), np.min(cols), np.max(cols)

def preprocess_test_images(df, src_img_dir, src_mask_dir, output_dir):
    if not os.path.exists(output_dir): os.makedirs(output_dir)
    reinhard = ReinhardNormalizer()
    valid_files = []
    
    print(f"⚙️ Precomputing Test images -> {output_dir}")
    
    for _, row in tqdm(df.iterrows(), total=len(df)):
        img_name = row['filename']
        mask_name = img_name.replace("img", "mask") if "img" in img_name else img_name.replace(".png", "_mask.png")
        
        img_p = os.path.join(src_img_dir, img_name)
        mask_p = os.path.join(src_mask_dir, mask_name)
        out_p = os.path.join(output_dir, img_name)
        
        if os.path.exists(out_p):
            valid_files.append(img_name); continue
            
        try:
            image = cv2.imread(img_p)
            if image is None: continue 
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
            
            if os.path.exists(mask_p):
                mask = cv2.imread(mask_p, cv2.IMREAD_GRAYSCALE)
                if mask is not None:
                    bbox = get_bbox_from_mask(mask)
                    if bbox: image = image[bbox[0]:bbox[1]+1, bbox[2]:bbox[3]+1]
            
            image = reinhard(image)
            Image.fromarray(image).resize((IMG_SIZE, IMG_SIZE), Image.Resampling.LANCZOS).save(out_p, "PNG")
            valid_files.append(img_name)
        except: continue
    return valid_files

class TissueDataset(Dataset):
    def __init__(self, dataframe, precomputed_dir, transform=None):
        self.df = dataframe
        self.precomputed_dir = precomputed_dir 
        self.transform = transform

    def __len__(self): return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img_name = row['filename']
        img_path = os.path.join(self.precomputed_dir, img_name)
        try: image = Image.open(img_path).convert("RGB")
        except: image = Image.new('RGB', (IMG_SIZE, IMG_SIZE))
        if self.transform: image = self.transform(image)
        return image, img_name

def get_transforms():
    return transforms.Compose([
        transforms.Resize((IMG_SIZE, IMG_SIZE)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
    ])

# --- 3. MODELLI ---
def get_model(arch, num_classes):
    if arch == 'b0':
        model = models.efficientnet_b0(weights=None)
        model.classifier[1] = nn.Linear(model.classifier[1].in_features, num_classes)
    elif arch == 'b1':
        model = models.efficientnet_b1(weights=None)
        model.classifier[1] = nn.Linear(model.classifier[1].in_features, num_classes)
    return model.to(DEVICE)

# --- 4. ENGINE ---
def generate_final_submission():
    print(f"\n🚀 Generazione Ensemble Pesato: {WEIGHT_B0*100}% B0 + {WEIGHT_B1*100}% B1")
    
    # Label Encoder
    train_df = pd.read_csv(PATHS['train_labels'])
    le = LabelEncoder()
    le.fit(train_df['label'])
    num_classes = len(le.classes_)
    
    # Dati Test
    test_files = sorted([f for f in os.listdir(PATHS['test_img']) if f.endswith('.png')])
    test_df = pd.DataFrame({'filename': test_files})
    valid_files = preprocess_test_images(test_df, PATHS['test_img'], PATHS['test_mask'], PRECOMPUTED_DIR_TEST)
    test_df = test_df[test_df['filename'].isin(valid_files)].reset_index(drop=True)
    
    loader = DataLoader(TissueDataset(test_df, PRECOMPUTED_DIR_TEST, get_transforms()), 
                        batch_size=BATCH_SIZE, shuffle=False, num_workers=2)
    
    # Caricamento Modelli
    models_b0 = []
    models_b1 = []
    
    print("Caricamento modelli B0...")
    for s in SEEDS:
        for f in range(N_FOLDS):
            path = f"b0_seed{s}_fold{f}.pth"
            if os.path.exists(path):
                m = get_model('b0', num_classes)
                m.load_state_dict(torch.load(path, map_location=DEVICE))
                m.eval()
                models_b0.append(m)
                
    print("Caricamento modelli B1...")
    for s in SEEDS:
        for f in range(N_FOLDS):
            path = f"b1_seed{s}_fold{f}.pth" # Assumiamo che B1 salvi con questo prefisso
            if os.path.exists(path):
                m = get_model('b1', num_classes)
                m.load_state_dict(torch.load(path, map_location=DEVICE))
                m.eval()
                models_b1.append(m)

    print(f"Modelli pronti: {len(models_b0)} B0, {len(models_b1)} B1")
    
    # Inferenza
    preds, fnames = [], []
    with torch.no_grad():
        for imgs, names in tqdm(loader):
            imgs = imgs.to(DEVICE)
            inputs = [imgs, torch.flip(imgs, [3]), torch.flip(imgs, [2]), torch.rot90(imgs, 1, [2, 3])]
            
            # Calcolo probabilità B0
            prob_b0 = torch.zeros((imgs.size(0), num_classes)).to(DEVICE)
            if models_b0:
                for m in models_b0:
                    for x in inputs:
                        prob_b0 += torch.softmax(m(x), dim=1)
                prob_b0 /= (len(models_b0) * 4)
            
            # Calcolo probabilità B1
            prob_b1 = torch.zeros((imgs.size(0), num_classes)).to(DEVICE)
            if models_b1:
                for m in models_b1:
                    for x in inputs:
                        prob_b1 += torch.softmax(m(x), dim=1)
                prob_b1 /= (len(models_b1) * 4)
            
            # MEDIA PESATA
            if models_b0 and models_b1:
                final_prob = (prob_b0 * WEIGHT_B0) + (prob_b1 * WEIGHT_B1)
            elif models_b0:
                final_prob = prob_b0
            else:
                final_prob = prob_b1
                
            preds.extend(le.inverse_transform(final_prob.argmax(1).cpu().numpy()))
            fnames.extend(names)
            
    pd.DataFrame({'sample_index': fnames, 'label': preds}).to_csv('submission_FINAL_ENSEMBLE.csv', index=False)
    print("\n✅ Submission finale generata: submission_FINAL_ENSEMBLE.csv")

if __name__ == "__main__":
    generate_final_submission()
```
