# Librerie 

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

print("librerie importate")
```

```python
def load_images_from_folder(folder):
    """
    Load and preprocess images from a specified folder.

    Args:
        folder (str): Path to the folder containing images

    Returns:
        np.ndarray: Array of preprocessed images with shape (N, H, W, C)
    """
    images = []

    # Iterate through files in the specified folder
    for filename in os.listdir(folder):
        img = cv2.imread(os.path.join(folder, filename))

        # Normalize image pixel values to a float range [0, 1]
        img = (img / 255).astype(np.float32)

        # Convert image from BGR to RGB
        img = img[...,::-1]

        # Make the image dataset squared
        dim = min(img.shape[:-1])
        img = img[(img.shape[0]-dim)//2:(img.shape[0]+dim)//2,
                  (img.shape[1]-dim)//2:(img.shape[1]+dim)//2, :]

        # Resize the image to 224x224 pixels
        img = cv2.resize(img, (224, 224))

        if img is not None:
            images.append(img)

    return np.array(images)

train_data_path = '/kaggle/input/an2dl-challenge2/an2dl/train_data/'
train_data = load_images_from_folder(train_data_path)

print(f"Loaded {len(train_data)} animal images")
```

```python
test_data_path = '/kaggle/input/an2dl-challenge2/an2dl/test_data/'
test_data = load_images_from_folder(test_data_path)

print(f"Loaded {len(test_data)} animal images")
```

```python
np.save('train_data.npy', train_data)
np.save('test_data.npy', test_data)

print("numpy arrays salvate")
```

# Esplorazione dei dati 

```python
# Number of images to display
num_img = 10

fig, axes = plt.subplots(2, num_img//2, figsize=(20, 9))
for i in range(num_img):
    ax = axes[i%2, i%num_img//2]
    ax.imshow(np.clip(train_data[i], 0, 1))  # Display clipped item images
    ax.axis('off')
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(2, num_img//2, figsize=(20, 9))
for i in range(num_img):
    ax = axes[i%2, i%num_img//2]
    ax.imshow(np.clip(test_data[i], 0, 1))  # Display clipped item images
    ax.axis('off')
plt.tight_layout()
plt.show()
```
